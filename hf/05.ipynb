```python
url = "https://mek.oszk.hu/11800/11864/html/index.html"
resp = requests.get(url, headers={"User-Agent": "Mozilla/5.0"})
resp.encoding = "utf-8"
html = resp.text
sel = Selector(text=html)

print("Article title:", sel.css("title::text").get())
```

```python
import scrapy
from scrapy.crawler import CrawlerRunner
from scrapy.utils.defer import deferred_to_future
from scrapy.utils.reactor import install_reactor

class SimpleSpider(scrapy.Spider):
    name = "my_spider"
    start_urls = ["https://mek.oszk.hu/11800/11864/html/index.html"]

    def parse(self, response):
        print("Page title:", response.css("title::text").getall())

runner = CrawlerRunner(settings={
    "LOG_LEVEL": "WARN",
    "TWISTED_REACTOR": "twisted.internet.asyncioreactor.AsyncioSelectorReactor",
})

await deferred_to_future(runner.crawl(SimpleSpider))
```
